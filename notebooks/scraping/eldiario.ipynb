```python
import feedparser
from bs4 import BeautifulSoup
import time
import pandas as pd
import html

import json
import os
```

```python

RSS_MAP = {
    "politica": "https://www.eldiario.es/rss/politica",
    "internacional": "https://www.eldiario.es/rss/internacional",
    "cultura": "https://www.eldiario.es/rss/cultura"}

def obtener_articulos(url, seccion, nombre_seccion, periodico, nombre_periodico):
    noticias=[]
    feed = feedparser.parse(url)

    for entry in feed.entries:

        noticia = {
            "Link": entry.link,
            "Periódico": nombre_periodico,
            "Fecha": time.strftime("%Y-%m-%d", entry.published_parsed),
            "Título": entry.title,
            "Subtítulo": "",
            "Categoría": nombre_seccion,
            "Contenido": html.unescape(entry.summary)
        }
                
        noticias.append(noticia)

    return noticias
```

```python
def guardar_noticias(noticias, archivo_json):
    
    # Si el archivo existe, cargar su contenido
    if os.path.exists(archivo_json):
        with open(archivo_json, "r", encoding="utf-8") as f:
            datos_existentes = json.load(f)
    else:
        datos_existentes = []

    # Añadir las nuevas noticias
    datos_existentes.extend(noticias)

    # Guardar todo de nuevo
    with open(archivo_json, "w", encoding="utf-8") as f:
        json.dump(datos_existentes, f, ensure_ascii=False, indent=4)
```

```python
ruta_guardado = "./../../data/elDiario.json"

noticias = obtener_articulos(url=RSS_MAP["politica"],
                             seccion="Politica", nombre_seccion="Nacional",
                             periodico="ElDiario", nombre_periodico="ElDiario")
guardar_noticias(noticias=noticias, archivo_json=ruta_guardado)

noticias = obtener_articulos(url=RSS_MAP["internacional"],
                             seccion="Internacional", nombre_seccion="Internacional",
                             periodico="ElDiario", nombre_periodico="ElDiario")
guardar_noticias(noticias=noticias, archivo_json=ruta_guardado)

noticias = obtener_articulos(url=RSS_MAP["cultura"],
                             seccion="Cultura", nombre_seccion="Cultura",
                             periodico="ElDiario", nombre_periodico="ElDiario")
guardar_noticias(noticias=noticias, archivo_json=ruta_guardado)
```
