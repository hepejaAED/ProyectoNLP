# Fake News en Youtube

````python
import pandas as pd

data_11s = pd.read_parquet('data_youtube/youtube_11s_enriched.parquet', engine='pyarrow')

data_cambio = pd.read_parquet('data_youtube/youtube_cambio_enriched.parquet', engine='pyarrow')

data_covid = pd.read_parquet('data_youtube/youtube_covid_enriched.parquet', engine='pyarrow')
````

````python
ranking_11s = pd.read_csv('data_youtube/rankings/ranking_consensus_11s.csv')

ranking_cambio = pd.read_csv('data_youtube/rankings/ranking_consensus_climate.csv')

ranking_covid = pd.read_csv('data_youtube/rankings/ranking_consensus_covid.csv')
````

````python
ranking_cambio.head()
````

````python
data_cambio.shape, ranking_cambio.shape
````

````python
df_11s = pd.merge(data_11s, ranking_11s[['video_id', 'mean_fake_prob']], on='video_id', how='inner')

df_cambio = pd.merge(data_cambio, ranking_cambio[['video_id', 'mean_fake_prob']], on='video_id', how='inner')

df_covid = pd.merge(data_covid, ranking_covid[['video_id', 'mean_fake_prob']], on='video_id', how='inner')
````

````python
df_11s.columns
````

````python
videos_11s = []

for i in range(df_11s.shape[0]):
    info_video = df_11s.iloc[i]

    video = {
        "Link": f"https://www.youtube.com/watch?v={info_video.video_id}",
        "Canal": info_video.channel,
        "Fecha": info_video.published_at[:10],
        "Título": info_video.title,
        "Subtítulo": info_video.description,
        "Categoría": "11s",
        "Contenido": info_video.transcript
    }

    videos_11s.append(video)
````

````python
videos_cambio = []

for i in range(df_cambio.shape[0]):
    info_video = df_cambio.iloc[i]

    video = {
        "Link": f"https://www.youtube.com/watch?v={info_video.video_id}",
        "Canal": info_video.channel,
        "Fecha": info_video.published_at[:10],
        "Título": info_video.title,
        "Subtítulo": info_video.description,
        "Categoría": "Cambio climático",
        "Contenido": info_video.transcript
    }

    videos_cambio.append(video)
````

````python
videos_covid = []

for i in range(df_covid.shape[0]):
    info_video = df_covid.iloc[i]

    video = {
        "Link": f"https://www.youtube.com/watch?v={info_video.video_id}",
        "Canal": info_video.channel,
        "Fecha": info_video.published_at[:10],
        "Título": info_video.title,
        "Subtítulo": info_video.description,
        "Categoría": "Covid",
        "Contenido": info_video.transcript
    }

    videos_covid.append(video)
````

````python
import json

ruta_archivo_11s = "data/youtube_11s.json"
ruta_archivo_cambio = "data/youtube_cambio.json"
ruta_archivo_covid = "data/youtube_covid.json"

with open(ruta_archivo_11s, 'w', encoding = 'utf-8') as archivo:
    json.dump(videos_11s, archivo, indent = 1, ensure_ascii = False)

with open(ruta_archivo_cambio, 'w', encoding = 'utf-8') as archivo:
    json.dump(videos_cambio, archivo, indent = 1, ensure_ascii = False)

with open(ruta_archivo_covid, 'w', encoding = 'utf-8') as archivo:
    json.dump(videos_covid, archivo, indent = 1, ensure_ascii = False)
````
